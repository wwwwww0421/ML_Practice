```python
from fraudlib.synth import make_transactions

df = make_transactions()
```

```python
df
```

```python
df.groupby("customer_id").agg("amount").count().sort_values(ascending=False).head(10)
```

```python
# df.groupby(["channel", "is_fraud"]).agg("amount").count().sort_values(ascending=False).head(10)

df[df['is_fraud'] == 1].groupby('channel').agg('amount').count().sort_values(ascending=False) / df[df['is_fraud'] == 0].groupby('channel').agg('amount').count().sort_values(ascending=False)

## We can see that fraud ratio are similar throughout all channels.
```

```python
fraud_by_id = df[df['is_fraud'] == 1].groupby('customer_id')['amount'].count()
non_fraud_by_id = df[df['is_fraud'] == 0].groupby('customer_id')['amount'].count()
ratio = (fraud_by_id / non_fraud_by_id)
# ratio.where(ratio > 0.5).sort_values(ascending=False).head(10)
ratio[ratio > 0.03].sort_values(ascending=False).head(20)

## Here are few started customers with high fraud ratio. We can investigate their transactions further to see if there are any patterns.
```

```python
df[(df['customer_id'].isin(ratio[ratio > 0.03].index)) & df['is_fraud'] == 1].groupby(['channel']).agg('amount').count().sort_values(ascending=False).head(20) / df[(df['customer_id'].isin(ratio[ratio > 0.03].index)) & df['is_fraud'] == 1].shape[0]

## We can see in high fraud rate customers, the fraud transactions are likely to be happened through app.
```
